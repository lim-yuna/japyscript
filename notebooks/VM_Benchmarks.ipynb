# JapyScript VM Benchmarks

Measures instruction dispatch throughput, garbage collection pauses, and
function-call overhead of the reference VM across representative programs.

```python
import sys, time, statistics
from pathlib import Path

sys.path.insert(0, str(Path('..').resolve()))
from compiler import compile_source
from vm import VM

print('python:', sys.version.split()[0])
```

```python
def bench(source, label, runs=5):
    samples = []
    for _ in range(runs):
        code = compile_source(source, '<bench>')
        t0 = time.perf_counter()
        VM().run(code)
        samples.append(time.perf_counter() - t0)
    print(f'{label:30s}  mean={statistics.mean(samples)*1000:7.2f}ms  p95={sorted(samples)[-1]*1000:7.2f}ms')
```

## Instruction dispatch

Tight arithmetic loop.

```python
bench('''
total = 0
for n in range(1000):
    total += n
''', 'arith loop 1000')
```

## Function call overhead

```python
bench('''
def inc(n): return n + 1
total = 0
for n in range(500):
    total = inc(total)
''', 'function call 500')
```

## String concatenation

```python
bench('''
s = ""
for n in range(200):
    s += "x"
''', 'string concat 200')
```
